```python
# Missão 1: O Banco de Dados (Dicionários e Listas)
# O objetivo aqui é criar o nosso "estoque" de heróis.

#lista - liga_da_justica.
liga_da_justiça = [
    {"nome": "Krypto",
     "poder": 98,
     "cidade": "inter"},

    {"nome": "Super-Homem",
     "poder": 83,
     "cidade": "Aveador"
     },

    {"nome": "Batmam",
     "poder": 65,
     "cidade": "gotham"
     }
]

# Alterando uma cidade especifica de um héroi do indice[0] = Krypto
liga_da_justiça[1]["cidade"] = "Metropolis"

#Adicionando um novo herói no dicinário ->  Xavier
novo_heroi = {"nome": "Xavier",
     "poder": 49,
     "cidade": "Mansão"
}
#Adicionando herói na lista
liga_da_justiça.append(novo_heroi)



# Higher-Order Functions(HOF).
def e_poderoso(pers):
 return pers["poder"] > 80

def mora_em_gotham(pers):
  return pers["cidade"].lower() == "gotham"

def heroi_auditor(lista_herois, criterio):
  print(f"\n---- lIGA DE HÉROIS: {criterio.__name__}----")
  for heroi in lista_herois:
    if criterio(heroi):
       print(heroi)

heroi_auditor(liga_da_justiça, e_poderoso)
heroi_auditor(liga_da_justiça, mora_em_gotham)

print(f"\n---- Treinamento -----")

# Closures(Funções de Fábrica)
def gerar_treinamento(bonus):
  def aplicar_trieno(heroi):
     heroi["poder"] += bonus
     return heroi
  return aplicar_trieno

treino_leve = gerar_treinamento(2)
treino_pesado = gerar_treinamento(36)


treino_leve(liga_da_justiça[2])
treino_leve(liga_da_justiça[0])
treino_leve(liga_da_justiça[3])

print("\n-=-=-=-=- TREINO LEVE -=-=-=-=- ")
print(f"Poder antes:\nHéroi: {liga_da_justiça[2]["nome"]} \nPoder: {liga_da_justiça[2]['poder']} ")
print()
print(f"Poder antes:\nHéroi: {liga_da_justiça[0]["nome"]} \nPoder: {liga_da_justiça[0]['poder']} ")
print()
print(f"Poder antes:\nHéroi: {liga_da_justiça[3]["nome"]} \nPoder: {liga_da_justiça[3]['poder']} ")

treino_pesado(liga_da_justiça[2])
treino_pesado(liga_da_justiça[0])
treino_pesado(liga_da_justiça[3])

print("\n-=-=-=-=- TREINO PESADO -=-=-=-=- ")
print(f"\nPoder depois do treino pesado: \nHéroi: {liga_da_justiça[2]["nome"]} \nPoder: {liga_da_justiça[2]['poder']}")
print()
print(f"\nPoder depois do treino pesado: \nHéroi: {liga_da_justiça[0]["nome"]} \nPoder: {liga_da_justiça[0]['poder']}")
print()
print(f"\nPoder depois do treino pesado: \nHéroi: {liga_da_justiça[3]["nome"]} \nPoder: {liga_da_justiça[3]['poder']}")


print("\nNova lista atualizada:")
for heroi in liga_da_justiça:
    print(heroi)
```
